```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso
from sklearn import metrics
from sklearn.model_selection import train_test_split



pd.set_option('future.no_silent_downcasting', True) 
```

```python
car_dataset= pd.read_csv("car data.csv")
```

```python
car_dataset
```

```python
car_dataset.shape
```

```python
car_dataset.info()
```

```python
car_dataset.isnull().sum()
```

```python
print (car_dataset.Fuel_Type.value_counts())
print (car_dataset.Seller_Type.value_counts())
print (car_dataset.Transmission.value_counts())
```

```python
car_dataset.replace({"Fuel_Type":{"Petrol":0,"Diesel":1,"CNG":2}},inplace= True)

car_dataset.replace({"Seller_Type":{"Dealer":0,"Individual":1}},inplace= True)

car_dataset.replace({"Transmission":{"Manual":0,"Automatic":1}},inplace= True)
```

```python
car_dataset.head()
```

```python
X = car_dataset.drop(["Car_Name","Selling_Price"],axis= 1)
```

```python
X
```

```python
Y = car_dataset["Selling_Price"]
```

```python
Y
```

```python
X_train,X_test,Y_train,Y_test= train_test_split(X,Y,test_size=0.1,random_state=2)
# stratify y uygun gruplara bölmeye calısıyor ama grupta en fazla bir üye varsa sorun cıkarır
# veri sayısı az oldug icin 0.1 aldık test size degerini 
```

```python
lin_reg_model = LinearRegression()
```

```python
lin_reg_model.fit(X_train,Y_train)
```

```python
training_data_prediction= lin_reg_model.predict(X_train)
```

```python
error_score = metrics.r2_score(Y_train,training_data_prediction)
```

```python
error_score
```

## gercek ve tahmin edilen degerleri karsılastırıcaz 

```python
plt.scatter(Y_train,training_data_prediction)
plt.xlabel("actual price ")
plt.ylabel("predict price")
plt.show
```

fiyatlar birbirine yakın renkleri aynı ama farklı değerler biri tahmin biri asıl değer vs 

```python
test_data_prediction= lin_reg_model.predict(X_test)
```

```python
error_score = metrics.r2_score(Y_test,test_data_prediction)
```

```python
error_score
```

```python
plt.scatter(Y_test,test_data_prediction)
plt.xlabel("actual price ")
plt.ylabel("predict price")
plt.show;
```
